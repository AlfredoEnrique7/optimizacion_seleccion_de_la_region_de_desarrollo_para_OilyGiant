## Proyecto de Optimización de Pozos Petroleros: Selección de la Región de Desarrollo para OilyGiant

### Introducción
En la industria de la exploración y extracción de petróleo, la selección de nuevas zonas de perforación implica inversiones multimillonarias de alto riesgo geológico y financiero. Para la compañía petrolera **OilyGiant**, maximizar el margen de ganancias depende directamente de la precisión con la que se puedan estimar las reservas antes de iniciar las operaciones de infraestructura. 

Este proyecto aplica técnicas avanzadas de **Ciencia de Datos y Machine Learning (ML)** para predecir el volumen de reservas en pozos de tres regiones geográficas distintas. A través de un enfoque basado en modelos predictivos y simulaciones estadísticas, se determina de manera rigurosa cuál de las regiones ofrece el mayor beneficio económico neto, mitigando el riesgo de pérdidas bajo las restricciones presupuestarias y operativas de la empresa.

---

### Objetivos del Proyecto

* **Desarrollar un modelo predictivo de ML:** Construir y validar un modelo de regresión lineal para cada una de las tres regiones geográficas, capaz de estimar el volumen de reservas de los pozos a partir de sus características geológicas (`f0`, `f1`, `f2`).
* **Establecer el punto de equilibrio financiero (*Break-Even*):** Calcular el volumen mínimo de producción por pozo requerido para cubrir los costos de inversión y desarrollo, comparándolo con los promedios reales de cada región.
* **Evaluar ganancias potenciales y riesgo mediante Bootstrap:** Implementar una simulación de Bootstrap con 1000 muestras aleatorias para calcular la ganancia promedio esperada, el intervalo de confianza del 95% y la probabilidad de pérdidas para cada territorio.
* **Mitigar el riesgo corporativo:** Seleccionar la región óptima para la perforación que garantice un riesgo de pérdida financiera inferior al **2.5%**, cumpliendo estrictamente con la política de gestión de riesgos de **OilyGiant**.

### 3. Limpieza de Datos e Ingeniería de Características (Feature Engineering)
Una vez diagramada la estructura inicial, es indispensable preparar los datos para cumplir de forma estricta con los supuestos matemáticos de la Regresión Lineal y las exigencias de calidad de los criterios de evaluación. Esta fase comprende un proceso depurado de curación de datos y selección de variables:

* **Tratamiento de Duplicados en Claves Únicas:** Aunque los datasets no presentan duplicaciones completas en todas sus filas, se audita la columna `id` (identificador único del pozo). La presencia de IDs repetidos con diferentes mediciones geológicas revela un riesgo de inconsistencia, por lo que se implementa un filtrado estricto para mantener únicamente el primer registro de cada pozo y posteriormente resetear los índices de forma limpia.
* **Selección de Características (Feature Selection):** La Regresión Lineal requiere exclusivamente variables numéricas continuas que aporten poder predictivo. Por lo tanto, se descarta la columna categórica `id`, ya que no posee relevancia geológica y actúa únicamente como metadato, aislando las variables predictoras (`f0`, `f1`, `f2`) del objetivo de estudio (`product`).
* **Ingeniería de Características:** Se estructuran los diccionarios y colecciones de datos que gobernarán la automatización del pipeline para las tres regiones, asegurando que las transformaciones y segmentaciones de entrenamiento/validación (proporción 75:25) se realicen de manera idéntica y reproducible.

```python
# =====================================================================
# LIMPIEZA DE DATOS E INGENIERÍA DE CARACTERÍSTICAS
# =====================================================================

print('\n======================================================')
print("REMOCION DE REGISTROS INCONSISTENTES (IDs DUPLIDADOS)")
print('======================================================')

for nombre, df in zip(regiones, datasets):
    # Contabilizar cuántas claves únicas de pozo se repiten en el muestreo
    duplicados_id = df.duplicated(subset=['id']).sum()
    print(f"🔸 {nombre} - Registros con ID de pozo duplicado detectados: {duplicados_id}")
    
    if duplicados_id > 0:
        # Se remueven los registros duplicados manteniendo la primera ocurrencia por seguridad
        df.drop_duplicates(subset=['id'], keep='first', inplace=True)
        # Se restablece el índice para evitar desalineaciones en la división Train/Valid
        df.reset_index(drop=True, inplace=True)
        print(f"   -> {nombre} - Registros limpios y consolidados: {df.shape[0]}")
```

### 4. Análisis de Correlación de Variables (Estadística exigida)
Antes de proceder con la segmentación de los datos y el entrenamiento de los algoritmos de ML, es un requisito metodológico crucial evaluar el grado de asociación lineal entre las variables predictoras del subsuelo (`f0`, `f1`, `f2`) y el volumen de reservas extraídas (`product`). Los coeficientes resultantes se visualizarán mediante mapas de calor (*heatmaps*), sirviendo como un diagnóstico previo indispensable para comprender el comportamiento geológico de cada territorio y anticipar el rendimiento que tendrá el modelo predictivo lineal propuesto.


```python
print('\n===============================================')
print("MATRICES DE CORRELACIÓN DE PEARSON GENERADAS")
print('===============================================')

# Configurar un lienzo unificado para comparar las 3 regiones visualmente
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
fig.suptitle('Matrices de Correlación de Pearson por Región Geográfica', fontsize=16, fontweight='bold')

for i, (nombre, df) in enumerate(zip(regiones, datasets)):
    # Aislar las variables numéricas continuas para el cálculo matemático de Pearson
    df_numerico = df[['f0', 'f1', 'f2', 'product']]
    matriz_corr = df_numerico.corr()
    
    # Renderizar el mapa de calor (Heatmap) con anotaciones numéricas explícitas
    sns.heatmap(matriz_corr, annot=True, cmap='RdBu', vmin=-1, vmax=1, fmt=".3f", ax=axes[i], cbar=True)
    axes[i].set_title(f'Mapa de Calor - {nombre}', fontweight='bold')

# Guardar automáticamente los gráficos de las Matrices de Pearson por región como imagen
plt.tight_layout()
plt.savefig('matriz_correlacion_regiones.png', dpi=300)

# Mostrar los gráficos de las Matrices de Pearson por región
plt.show()
```

### 5. Entrenamiento y Validación del Modelo
Una vez consolidada la calidad de los datos y aisladas las variables predictoras, se inicia la fase de modelado estadístico mediante aprendizaje supervisado. El criterio de evaluación de la plataforma exige la implementación estricta del algoritmo de **Regresión Lineal**, un modelo paramétrico ideal para este escenario debido a su interpretabilidad y a las relaciones lineales identificadas previamente en el análisis de correlación. Para cada una de las tres regiones geográficas, el conjunto de datos se divide de forma rigurosa en una proporción de **75% para el entrenamiento** del modelo y **25% para su validación**. El objetivo central de esta sección es ajustar los coeficientes del algoritmo, generar las predicciones del volumen de reservas de petróleo para los conjuntos de prueba y calcular la métrica de rendimiento obligatoria: la **Raíz del Error Cuadrático Medio (RMSE)**. Esto permite cuantificar la precisión del modelo y evaluar la dispersión de los errores antes de proceder con los cálculos de rentabilidad financiera.


```python
# =====================================================================
# CONSTANTES DEL PROYECTO (Ajustadas a dólares USD según la consigna)
# =====================================================================
PRESUPUESTO = 100000000          # 100 millones de dólares USD
PUNTOS_EXPLORACION = 500         # Puntos investigados al explorar
PUNTOS_DESARROLLO = 200          # Mejores puntos seleccionados para desarrollo
VALOR_UNIDAD_PRODUCTO = 4500     # Ingreso por cada unidad de 'product' (4,500 USD)
LIMITE_RIESGO_PERDIDA = 0.025    # Riesgo máximo permitido (2.5%)

# ==========================================
# ENTRENAMIENTO Y EVALUACIÓN DEL MODELO
# ==========================================

print('\n==================================================')
print("ENTRENANDO MODELOS DE REGRESIÓN LINEAL")
print('====================================================')

# Diccionarios para almacenar predicciones y valores reales para el análisis posterior
valores_reales = {}
valores_predichos = {}

for i, df in enumerate(datasets):
    # Separar características (X) y objetivo (y). Se excluye 'id' por no ser numérica
    X = df[['f0', 'f1', 'f2']]
    y = df['product']
    
    # División del dataset: 75% entrenamiento, 25% validación (Exigencia estricta)
    X_train, X_valid, y_train, y_valid = train_test_split(X, y, test_size=0.25, random_state=42)
    
    # Inicializar y entrenar el modelo de Regresión Lineal
    model = LinearRegression()
    model.fit(X_train, y_train)
    
    # Realizar predicciones en el set de validación
    preds = model.predict(X_valid)
    
    # Convertir a Series de Pandas con índices alineados para evitar desalineación en Bootstrap
    valores_reales[i] = y_valid.reset_index(drop=True)
    valores_predichos[i] = pd.Series(preds)
    
    # Calcular volumen promedio predicto y RMSE
    volumen_promedio_predicho = valores_predichos[i].mean()
    rmse = mean_squared_error(valores_reales[i], valores_predichos[i]) ** 0.5
    
    print(f"Región {i} -> Volumen Promedio Predicho: {volumen_promedio_predicho:.2f} | RMSE: {rmse:.2f}")
```

#### Análisis y Conclusiones del Modelado Predictivo
Tras entrenar los modelos individuales de Regresión Lineal para cada región geográfica con la proporción de datos 75:25, los resultados obtenidos en el conjunto de validación revelan un comportamiento técnico altamente contrastante que define la estrategia del proyecto:

* **Región 0 (RMSE: `37.69` | Volumen Predicho: `92.61` u):** Esta región presenta un volumen promedio estimado bastante alto. Sin embargo, su elevado RMSE (casi el 40% del valor promedio de las reservas) demuestra que las predicciones del modelo tienen una dispersión de error masiva. Existe una alta incertidumbre geológica en este territorio, lo que disminuye la confiabilidad de las proyecciones.
* **Región 1 (RMSE: `0.89` | Volumen Predicho: `68.58` u):** Aunque esta región registra el promedio de reservas más bajo, el comportamiento del modelo es excepcional. El análisis del **punto de equilibrio (*Break-Even*)** permite calcular el volumen mínimo de reservas de petróleo (`product`) que un pozo individual debe contener para justificar los costos de infraestructura, perforación y desarrollo sin incurrir en pérdidas económicas. 

A partir de los parámetros financieros fijados por la compañía —un presupuesto global de **100 millones de dólares (USD)** para el desarrollo de 200 pozos y un ingreso de **4,500 dólares** por cada unidad de producto extraído— se establecerá el umbral de rentabilidad matemática. El objetivo de esta sección es comparar este límite crítico contra los volúmenes de reservas promedio reales de cada región geográfica, sirviendo como fundamento económico para evaluar si la estrategia de perforación requiere obligatoriamente el filtrado inteligente mediante algoritmos de Machine Learning o si las regiones son rentables por sí mismas.

### 7. Cálculo de Ganancias y Distribución de Riesgos (Análisis Inicial y Bootstrap)

Tras determinar el punto de equilibrio financiero, la fase final y más crítica del proyecto consiste en evaluar los beneficios económicos potenciales de cada región y mitigar la incertidumbre del subsuelo para la toma de decisiones corporativas. Esta fase se estructurará en dos enfoques analíticos complementarios:

#### Enfoque Determinista (Análisis Inicial Completo)
Antes de simular escenarios de incertidumbre, se realizará una **comparación independiente directa sobre los conjuntos de validación completos de cada región**. Utilizando la función de ganancias desarrollada, se seleccionarán de forma idealizada los 200 pozos con las mayores estimaciones previstas por los modelos y se calculará su volumen real total de reservas y su beneficio potencial. Este paso es fundamental para identificar qué región se perfila inicialmente como la más lucrativa bajo condiciones ideales de información completa.

#### Enfoque Probabilístico (Simulación Bootstrap)
Posteriormente, se contrastará el análisis inicial implementando la técnica estadística de **Bootstrap mediante un proceso iterativo de 1,000 submuestras**. En la operación real en campo, OilyGiant no puede perforar directamente los mejores pozos de un lote masivo; primero debe explorar una muestra aleatoria limitada de 500 puntos para luego seleccionar y desarrollar los mejores 200 pozos de ese subconjunto. 

La simulación de Bootstrap replicará esta variabilidad del muestreo de campo, permitiendo al modelo de Regresión Lineal preseleccionar los pozos en cada escenario, mientras que las ganancias netas reales se calcularán con los valores verdaderos del subsuelo (`target`). El objetivo principal es construir una distribución de probabilidad de las ganancias por región para calcular la ganancia promedio esperada, el intervalo de confianza del 95% (percentiles 2.5% y 97.5%) y la probabilidad de pérdida financiera. Este análisis proveerá el criterio científico definitivo para identificar cuál región cumple con la política de OilyGiant de mantener un riesgo de pérdida inferior al **2.5%**.


```python
# ==========================================
# FUNCIÓN PARA EL CÁLCULO DE GANANCIAS
# ==========================================

def calcular_ganancia(target, predictions, count):
    """
    Selecciona los mejores pozos basados en las predicciones del modelo,
    pero calcula la ganancia real utilizando los valores verdaderos (target).
    """
    preds_ordenadas = predictions.sort_values(ascending=False)
    mejores_targets = target[preds_ordenadas.index][:count]
    ingreso_total = mejores_targets.sum() * VALOR_UNIDAD_PRODUCTO
    ganancia_neta = ingreso_total - PRESUPUESTO
    return ganancia_neta

# =====================================================================
# 7. COMPARACIÓN INICIAL DE GANANCIAS EN EL SET DE VALIDACIÓN COMPLETO
# =====================================================================

print(f"=================================================================")
print("Análisis Inicial de Ganancias (Top 200 Pozos por Región)")
print(f"=================================================================\n")

for i in range(3):
    target = valores_reales[i]
    preds = valores_predichos[i]
    
    # Calculamos la ganancia ideal usando el set de validación completo
    ganancia_inicial = calcular_ganancia(target, preds, PUNTOS_DESARROLLO)
    
    # Calculamos el volumen real acumulado de esos 200 pozos
    preds_ordenadas = preds.sort_values(ascending=False)
    mejores_targets = target[preds_ordenadas.index][:PUNTOS_DESARROLLO]
    volumen_real_total = mejores_targets.sum()
    
    print(f"================ REGION {i} ================")
    print(f"Volumen real total de reservas (200 mejores pozos): {volumen_real_total:.2f} unidades")
    print(f"Ganancia potencial inicial calculada: ${ganancia_inicial:,.2f} USD")
```

#### Evaluación Inicial de Beneficios y Justificación Geológica

Al aplicar la función de ganancia sobre los conjuntos de validación completos para seleccionar los 200 pozos con las estimaciones más altas de cada región de forma directa (escenario ideal), se obtienen los siguientes resultados financieros reales:

*   **Región 0:** Registra un volumen acumulado de reservas de **29,930.07 unidades**, generando la ganancia potencial inicial más alta con **\$34,685,297.87 USD**.
*   **Región 1:** Registra un volumen acumulado de reservas de **27,589.08 unidades**, generando una ganancia potencial inicial de **\$24,150,866.97 USD**.
*   **Región 2:** Registra un volumen acumulado de reservas de **27,548.03 unidades**, generando una ganancia potencial inicial de **$23,966,130.01 USD**.

#### Justificación de Selección Inicial (Escenario Ideal)
Si la toma de decisiones se realizara exclusivamente bajo este análisis determinista inicial, **la empresa elegiría invertir en la Región 0**. Esta opción supera a la Región 1 por más de 10.5 millones de dólares en ganancias netas ideales y posee el mayor volumen acumulado de reservas reales en sus mejores 200 pozos seleccionados.

#### El Dilema de la Incertidumbre: Por qué esta elección cambia con el Bootstrap?
Esta selección inicial representa un **escenario idealizado e irreal**, ya que asume de manera falsa que la empresa conoce de antemano el comportamiento de todo el territorio y puede ir de forma directa a perforar los 200 mejores pozos de entre las 25,000 locaciones disponibles en el set de validación.

En la operación real en campo, OilyGiant solo puede explorar una muestra aleatoria limitada de 500 puntos para seleccionar y desarrollar los mejores 200 de ese lote específico. Al incorporar esta restricción estadística mediante la técnica de **Bootstrap** en la siguiente sección, la conclusión da un giro radical: debido a que los modelos matemáticos de las Regiones 0 y 2 tienen un margen de error muy alto (RMSE > 37), seleccionarán "falsos positivos" (pozos que parecían ricos en la predicción pero están vacíos en la realidad), lo que causará que sus riesgos de pérdida real en campo se disparen al **5.90% y 8.00%**. En cambio, la **Región 1**, gracias a su precisión extrema (RMSE: 0.89), filtrará el terreno de forma infalible, consolidándose como la única opción segura (Riesgo: 1.60%) y verdaderamente rentable para la corporación.

```python
# ================================================
# CÁLCULO DE RIESGOS Y GANANCIAS CON BOOTSTRAP
# ================================================
print('\n==================================================')
print("BOOTSTRAP EJECUTADO  (1,000 MUESTRAS POR REGIÓN)")
print('====================================================')
state = np.random.RandomState(42)

for i in range(3):
    ganancias_bootstrap = []
    target = valores_reales[i]
    preds = valores_predichos[i]
    
    # Simulación de 1,000 iteraciones
    for j in range(1000):
        # Muestrear 500 puntos aleatorios con reemplazo (Simula la exploración de un nuevo lote)
        target_subsample = target.sample(n=PUNTOS_EXPLORACION, replace=True, random_state=state)
        preds_subsample = preds[target_subsample.index]
        
        # Calcular la ganancia de los mejores 200 pozos dentro de esa muestra
        ganancia = calcular_ganancia(target_subsample.reset_index(drop=True), 
                                    preds_subsample.reset_index(drop=True), 
                                    PUNTOS_DESARROLLO)
        ganancias_bootstrap.append(ganancia)
        
    ganancias_bootstrap = pd.Series(ganancias_bootstrap)
    
    # Estadísticas clave solicitadas por la plataforma
    ganancia_promedio = ganancias_bootstrap.mean()
    intervalo_inferior = ganancias_bootstrap.quantile(0.025)
    intervalo_superior = ganancias_bootstrap.quantile(0.975)
    
    # Calcular el porcentaje de escenarios donde hubo pérdidas (ganancia < 0)
    probabilidad_perdida = (ganancias_bootstrap < 0).mean()
    
    print(f"\n================ REGION {i} ================")
    print(f"Ganancia Promedio Esperada: {ganancia_promedio:,.2f}")
    print(f"Intervalo de Confianza del 95%: [{intervalo_inferior:,.2f} , {intervalo_superior:,.2f}]")
    print(f"Riesgo de Pérdidas: {probabilidad_perdida * 100:.2f}%")
    
    if probabilidad_perdida < LIMITE_RIESGO_PERDIDA:
        print(" VERDICTO: Cumple con el criterio de riesgo. Región sugerida para inversión.")
    else:
        print(" VERDICTO: RECHAZADA. El riesgo supera el límite del 2.5%.")
```

#### Análisis Estadístico de Riesgos y Ganancias (Evaluación de Bootstrap)

Tras ejecutar la simulación de Bootstrap con 1,000 muestras aleatorias para replicar las condiciones reales de exploración y desarrollo de la compañía, se obtuvieron métricas probabilísticas definitivas que resuelven el problema de inversión:

*   **Región 0:** Reporta una ganancia promedio esperada de **`$4,035,720.64 USD`**. No obstante, su intervalo de confianza del 95% abarca pérdidas severas, iniciando en terreno negativo (**`-$1,376,900.03 USD`**). Esto se traduce en un **riesgo de pérdidas del 5.90%**, una cifra que supera con creces el límite corporativo tolerado por la empresa.
*   **Región 1:** Ofrece el mejor rendimiento financiero global con la **ganancia promedio esperada más alta del proyecto: `$4,393,636.70 USD`**. Es la única zona cuyo intervalo de confianza del 95% es **completamente positivo**, garantizando beneficios de entre `$317,477.00 USD` y `$8,458,750.96 USD`. Adicionalmente, presenta un **riesgo de pérdidas de apenas el 1.60%**, situándose holgadamente por debajo del límite estricto de la compañía.
*   **Región 2:** Genera una ganancia promedio esperada de **`$3,677,019.27 USD`**. Al igual que la Región 0, su intervalo de confianza expone a la empresa a números rojos en los escenarios pesimistas (**`-$1,691,885.59 USD`**) y registra el **riesgo de pérdidas más adverso del estudio con un 8.00%**.
